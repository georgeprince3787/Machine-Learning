```python
import sqlite3 as sq
connections = sq.connect("example.db")
connections
cursor = connections.cursor()
```

```python
### Create A Table
cursor.execute("""
    CREATE TABLE IF NOT EXISTS employees(
        id INTEGER PRIMARY KEY,
        name TEXT NOT NULL,
        age INTEGER,
        department TEXT
   )
""")

## Commit the Changes
connections.commit()
cursor.execute("""
    SELECT * FROM  employees
    """)

```

```python
## Insert data in sqlite table
cursor.execute("""
    Insert Into employees(name,age,department)
               values("Raunak",32,"Data Scientist")

""")
cursor.execute("""
    Insert Into employees(name,age,department)
                values("Rahul",28,"Data Analyst")
""")
cursor.execute("""
    Insert Into employees(name,age,department)
                values("Rohit",30,"Data Engineer")
""")

## Commit the changes
connections.commit()
```

```python
### Query the data From the Table 
cursor.execute("Select * From employees")
rows = cursor.fetchall()
```

```python
### Update the Data in the Table
cursor.execute("""
    Update employees
    set age = 35
    where name = "Raunak"      
""")
connections.commit()

### Query the data From the Table 
cursor.execute("Select * From employees")
rows = cursor.fetchall()

## Print the quired data 
for row in rows:
    print(row)

connections.close()
```

## Working With Sales Data

```python
import sqlite3 as sq
## Connect to an SQLITE database
connection = sq.connect("sales_data.db")
cursor = connection.cursor()
```

```python
## Create a TABLE of sales Data 
cursor.execute("""
    CREATE TABLE IF NOT EXISTS sales(
        id TEXT PRIMARY KEY,
        date TEXT NOT NULL,
        product TEXT NOT NULL,
        sales INTEGER,
        region TEXT
    )
""")
```

```python
# Clear old data (prevents UNIQUE constraint error)
cursor.execute("DELETE FROM sales")

## Insert Data into the Sales Table
sales_data = [
    ("id-001","2024-01-01","Product A",100,"North"),
    ("id-002","2024-01-02","Product B",150,"South"),
    ("id-003","2024-01-03","Product C",200,"East"),
    ("id-004","2024-01-04","Product D",250,"West"),
    ("id-005","2024-01-05","Product E",300,"North"),
    ("id-006","2024-01-06","Product F",350,"South"),
    ("id-007","2024-01-07","Product G",400,"East"),
    ("id-008","2024-01-08","Product H",450,"West"),
    ("id-009","2024-01-09","Product I",500,"North"),
    ("id-010","2024-01-10","Product J",550,"South")
]

connection.executemany("""
    Insert Into sales(id,date,product,sales,region)
    values(?,?,?,?,?)
""",sales_data)

## Commit the changes
connection.commit()

## close the connection
connection.close()
```
