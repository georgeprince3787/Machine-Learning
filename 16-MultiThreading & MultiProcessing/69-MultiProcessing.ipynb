```python
import multiprocessing
import time
```

```python
def square_number(n):
    for i in range(5):
        time.sleep(1)
        print(f"Process {n}: {i}^2 = {i**2}")
```

```python
def cube_number(n):
    for i in range(5):
        time.sleep(1.5)
        print(f"Process {n}: {i}^3 = {i**3}")
```

```python
if __name__ == "__main__":
    p1 = multiprocessing.Process(target=square_number)
    p2 = multiprocessing.Process(target=cube_number)

    t = time.time()
    ## Start the processes
    p1.start()
    p2.start()
    ## Wait for the processes to finish
    p1.join()
    p2.join()

    finish_time = time.time()
    print(f"Execution Time : {finish_time - t} seconds")
```
