```python
import threading
import time
```

```python
def print_numbers():
    for i in range(1,6):
        
        print(f"Number : {i}")
        time.sleep(1)
```

```python
def print_letters():
    for letter in "abcde":
        
        print(f"Letter : {letter}")
        time.sleep(1)
```

```python
t = time.time()
```

```python
print_numbers()
print_letters()
```

```python
finish_time = time.time()
print(f"Execution Time : {finish_time - t} seconds")
```

```python
## Create Two Threads
t1 = threading.Thread(target = print_numbers)
t2 = threading.Thread(target = print_letters)
t = time.time()

## Start the thread
t1.start()
t2.start()

## Wait for the thread to complete
t1.join()
t2.join()

finish_time = time.time()
print(f"Execution Time : {finish_time - t} seconds")
```
