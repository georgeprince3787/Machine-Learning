```python
## Dataset(Tips Dataset)
import seaborn as sns
df = sns.load_dataset('tips')
df.head()
```

```python
## Feature Encodeing [[Abel Encoding And One Hot Encodeing]]
## Independent and department features
X = df[['tip', 'sex', 'smoker', 'time', 'size']]
y = df['total_bill']
```

```python
## train test split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.25,random_state=10)
X_train.head()
```

```python
## Feature Encoding(label and Ohe Encodeing)
from sklearn.preprocessing import LabelEncoder
le1 = LabelEncoder()
le2 = LabelEncoder()
le3 = LabelEncoder()
```

```python
X_train['sex'] = le1.fit_transform(X_train['sex'])
X_train['smoker'] = le2.fit_transform(X_train['smoker'])
X_train['time'] = le3.fit_transform(X_train['time'])
```

```python
X_test['sex'] = le1.transform(X_test['sex'])
X_test['smoker'] = le2.transform(X_test['smoker'])
X_test['time'] = le3.transform(X_test['time'])
```

```python
X_test.head()
```

```python
## Onehot encoding---- Column Transformr
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
ct = ColumnTransformer(transformers=[('onehot',OneHotEncoder(drop='first'),[3])],remainder='passthrough')
```

```python
ct.fit_transform(X_train)
```

```python
import sys 
import numpy 
numpy.set_printoptions (threshold=sys.maxsize)
X_test = ct.fit_transform(X_test)
X_test
```

```python
## SVR---Support Vector Regression 
from sklearn.svm import SVR
svr = SVR()
svr.fit(X_train,y_train)
y_pred = svr.predict(X_test)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error
print(r2_score(y_test,y_pred))
print(mean_absolute_error(y_test,y_pred))
```

```python
## Hyper Parameter Tuining using GridSearchCV
from sklearn.model_selection import GridSearchCV

# defining parameter range
param_grid = {'C':[0.1,1,10,100,1000],'gamma':[1,0.1,0.01,0.001,0.0001],'kernel':['rbf']}
grid = GridSearchCV(SVR(),param_grid,refit=True,verbose=3)
```

```python
## fitting the model from grid search 
grid.fit(X_train,y_train)
grid.best_params_
grid_prediction = grid.predict(X_test)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error
print(r2_score(y_test,grid_prediction))
print(mean_absolute_error(y_test,grid_prediction))
```
