```python
from sklearn.datasets import make_classification
#make_classification()
```

```python
X,y = make_classification(n_samples=1000,n_redundant=0, n_features=20, n_classes=2, weights=[0.9, 0.1], random_state=12)
```

```python
import pandas as pd
cols = [f"f{i}" for i in range(20)]
df1 = pd.DataFrame(X, columns=cols)
df2 = pd.DataFrame(y, columns=["target"])
final_df = pd.concat([df1, df2], axis=1)
final_df.head()
```

```python
final_df["target"].value_counts()
```

```python
import matplotlib.pyplot as plt
plt.scatter(final_df["f0"], final_df["f1"], c=final_df["target"])
plt.xlabel("f0")
plt.ylabel("f1")
```

```python
## Transform the dataset
oversample = SMOTE()
X,y = oversample.fit_resample(final_df.drop("target", axis=1), final_df["target"])
X.shape, y.shape
y == 0
y[y==0]
len(y[y==0])
len(y[y==1])
```
