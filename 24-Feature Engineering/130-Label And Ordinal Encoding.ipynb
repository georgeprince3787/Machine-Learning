## Label Encoding

```python
import pandas as pd
df = pd.DataFrame({
    'Color': ['Red', 'Blue', 'Green', 'Green','Red','Blue']

})
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder
lbl_encoder = LabelEncoder()
lbl_encoder.fit_transform(df[['Color']])
```

```python
lbl_encoder.transform([['Blue']])
lbl_encoder.transform([['Green']])
lbl_encoder.transform([['Red']])
```

## Ordinal Encoding

```python
from sklearn.preprocessing import OrdinalEncoder
## Create a sample DataFrame with an ordinal feature
df = pd.DataFrame({
    'Size': ['Small', 'Medium', 'Large', 'Medium', 'Small']
})

df
```

```python
## Create an instance of OrdinalEncoder and then perform fit and transform
encoder = OrdinalEncoder(categories=[['Small', 'Medium', 'Large']])
encoded = encoder.fit_transform(df[['Size']])
encoder.transform([['Medium']])
encoder.transform([['Large']])
encoder.transform([['Small']])
```
