## Upsampling

```python
import numpy as np
import pandas as pd
```

```python
## Set the random seed for reproducibility
np.random.seed(123)
```

```python
## Create a dataFrame with two classes
n_samples = 1000
class_0_ratio = 0.9
n_class_0 = int(n_samples * class_0_ratio)
n_class_1 = n_samples - n_class_0
```

```python
n_class_0, n_class_1
```

```python
## Create my DataFrame with imbalanced dataset
class_0 = pd.DataFrame({
    'feature1': np.random.normal(loc=0, scale=1, size=n_class_0),
    'feature2': np.random.normal(loc=0, scale=1, size=n_class_0),
    'target': [0]*n_class_0
})

class_1 = pd.DataFrame({
    'feature1': np.random.normal(loc=1, scale=1, size=n_class_1),
    'feature2': np.random.normal(loc=1, scale=1, size=n_class_1),
    'target': [1]*n_class_1
})
```

```python
df = pd.concat([class_0,class_1]).reset_index(drop=True)
```

```python
df.head()
```

```python
df.tail()
```

```python
df["target"].value_counts()
```

```python
df[df["target"] == 1]
```

```python
df_minority = df[df["target"] == 1]
df_majority = df[df["target"] == 0]
```

```python
! pip install scikit-learn
```

```python
from sklearn.utils import resample
df_minority_upsampled = resample(df_majority,
                                 replace=True,# Sample with replacement
                                    n_samples=len(df_minority),# To match minority class
                                    random_state = 42 # Reproducible results
)
```

```python
df_minority_upsampled.head()
```

```python
df_upsampled = pd.concat([df_majority, df_minority_upsampled])
df_upsampled["target"].value_counts()
```

## Down Sampling

```python
import numpy as np
import pandas as pd
## Set the random seed for reproducibility
np.random.seed(123)
## Create a dataFrame with two classes
n_samples = 1000
class_0_ratio = 0.9
n_class_0 = int(n_samples * class_0_ratio)
n_class_1 = n_samples - n_class_0
```

```python
n_class_0, n_class_1
```

```python
## Create my DataFrame with imbalanced dataset
class_0 = pd.DataFrame({
    'feature1': np.random.normal(loc=0, scale=1, size=n_class_0),
    'feature2': np.random.normal(loc=0, scale=1, size=n_class_0),
    'target': [0]*n_class_0
})

class_1 = pd.DataFrame({
    'feature1': np.random.normal(loc=1, scale=1, size=n_class_1),
    'feature2': np.random.normal(loc=1, scale=1, size=n_class_1),
    'target': [1]*n_class_1
})

df = pd.concat([class_0,class_1]).reset_index(drop=True)
df.head()
df.tail()
df["target"].value_counts()
df[df["target"] == 1]
df_minority = df[df["target"] == 1]
df_majority = df[df["target"] == 0]
```

```python
from sklearn.utils import resample
df_majority_downsampled = resample(df_majority,
                                 replace=False, # sample without replacement
                                    n_samples=len(df_minority), # to match minority class
                                    random_state=42 # reproducible results
)
df_majority_downsampled.head()
```

```python
df_majority_downsampled.shape
```

```python
df_majority_downsampled = pd.concat([df_minority, df_majority_downsampled])
df_majority_downsampled = df_majority_downsampled.reset_index(drop=True)
df_majority_downsampled["target"].value_counts()
```

DownSampling Is Bad! Because we are loosing data points we can use upsampling
