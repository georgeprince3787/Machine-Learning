```python
from flask import Flask, render_template

app = Flask(__name__)

@app.route("/success/<int:score>")
def success(score):
    if score >= 50:
        result = "Passed"
    else:
        result = "Failed"
    return render_template("result.html", result=result)

if __name__ == "__main__":
    app.run(debug=True, use_reloader=False)
```
