```python
from flask import Flask
## WSGI Application
app = Flask(__name__)
@app.route('/')
def welcome():
    return "Welcome to Flask Framework"

if __name__ == "__main__":
    app.run(debug=True, use_reloader=False)
```

```python
from flask import Flask
app = Flask(__name__)
@app.route('/')
def welcome():
    return "Welcome to Flask Framework"

@app.route('/index')
def index():
    return "<html><h1>Welcome to Index Page</h1></html>"

if __name__ == "__main__":
    app.run(debug = True, use_reloader=False)
```
