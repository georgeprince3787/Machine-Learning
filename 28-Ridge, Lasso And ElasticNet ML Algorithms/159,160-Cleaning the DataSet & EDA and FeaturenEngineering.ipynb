```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 
import seaborn as sns
%matplotlib inline
```

```python
dataset = pd.read_csv("159.2-Algerian_forest_fires_dataset_UPDATE.csv",header= 1)
dataset.head()
```

###  Data Cleaning

```python
## Missing Values
dataset[dataset.isnull().any(axis=1)]
```

```python
df = dataset
df.loc[:122,"Region"] = 0
df.loc[122:,"Region"] = 1
df.info()
df[["Region"]] = df[["Region"]].astype(int)
df.head()
```

```python
df.isnull().sum()
```

```python
## Removing The Null Values
df = df.dropna().reset_index(drop= True)
df.head()
df.isnull().sum()
df.iloc[[122]]
```

```python
## Removing The 122nd row
df = df.drop(122).reset_index(drop= True)
df.columns
```

```python
## Fix spaces in columns names
df.colums = df.columns.str.strip()
df.columns
df.info()
```

```python
df.columns
```

```python
## Change the required columns as integer datatype
df[['day', 'month', 'year', 'Temperature', ' RH', ' Ws']] = df[['day', 'month', 'year', 'Temperature', ' RH', ' Ws']].astype(int)
```

```python
df.columns = df.columns.str.strip()
```

```python
## Changing Other Columns to other datatypes
objects = [features for features in df.columns if df[features].dtype == "O"]

for i in objects:
    if i != "Classes":
        try:
            df[i] = df[i].astype(float)
        except Exception as e:
            print(f"Error in column: {i}")
            print(e)
```

```python
## Lets save the cleaned dataset
df.to_csv("159.1-Algerian_forest_fires_cleaned_dataset.csv",index= False)
```

```python
## Explotary Data Analysis
# Drop day,month,year
df_copy = df.drop(["day","month","year"],axis= 1)
df_copy.head()
```

```python
## Encoding the Categories in Classes
df_copy['Classes'] = np.where(df_copy['Classes'].str.contains("not fire"),0,1)
df_copy.head()
```

```python
df_copy.tail()
```

```python
df_copy['Classes'].value_counts()
```

```python
## Plot density plot for all features
plt.style.use("seaborn-v0_8")
df_copy.hist(bins= 50,figsize= (20,15))
plt.show()
```

```python
## Percentage for Pie Chart
percentage = df_copy['Classes'].value_counts(normalize= True)*100
percentage
```

```python
## Plting Pie Chart 
classlabels= ["Fire","Not Fire"]
plt.figure(figsize= (12,7))
plt.pie(percentage,labels= classlabels,autopct= "%1.1f%%")
plt.title("Pie Chart Of Classes")
plt.show()
```

```python
import seaborn as sns

## Correlation 
corr_matrix = df_copy.corr()

sns.heatmap(corr_matrix, annot=True, cmap="coolwarm")
plt.show()
```
