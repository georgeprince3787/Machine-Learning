```python
import seaborn as sns
### Basic Ploting with Seaborn
sns.load_dataset("tips")

## Create a Scatter Plot
import matplotlib.pyplot as plt
tips = sns.load_dataset("tips")
sns.scatterplot(x = "total_bill",y = "tip",data = tips,color = "red")
plt.title("Scatter Plot For Total Bill VS tip")
plt.show()
```

```python
## Line PLot
sns.lineplot(x = "size",y = "total_bill",data = tips,color = "blue")
plt.title("Line Plot Between total_bii VS tip")
plt.show()
```

Catagrical Plot


```python
## Bar Plot
sns.barplot(x = "day",y = "total_bill", data = tips)
plt.title("Bar Plot Between Toatal Bill VS tip")
plt.show()
```

```python
###Box Plot
sns.boxplot(x = "day", y = "total_bill",data = tips)
```

```python
## Voilim PLot
sns.violinplot(x="day",y = "total_bill",data = tips, color = "orange")
plt.title("Voilin Plot Between Total Bill vs Tip")
plt.show()
```

```python
## Histograms
sns.histplot(tips["total_bill"],bins = 10,kde = True)
```

```python
## KDE plot
sns.kdeplot(tips["total_bill"],fill = True)
```

```python
## Pair Plot
sns.pairplot(tips)
```

```python
### Heat Map
corr = tips[["total_bill","tip","size"]].corr()
corr
sns.heatmap(corr,annot = True,cmap = "coolwarm")
```
