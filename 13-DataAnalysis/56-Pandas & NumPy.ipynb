```python
import pandas as pd

df = pd.read_csv("sales_data.csv")
## Fetch 1st 5 Rows
df.head(5)
df.tail(5)
df.describe()
df.dtypes
```

```python
## Handling Missing Values
df.isnull()
df[df.isnull()]
df.isnull().any(axis = 1)
df.isnull().sum()
df.fillna(0)
df_filled = df.fillna(0)
df_filled
```

```python
## Filling Missing Values with the Mean of the Column
df["Sales_Fill_NA"] = df["Sales"].fillna(df['Sales'].mean())
df
```

```python
### Renaming Columns
df = df.rename(columns = {"Sales" : "Sales_data"})
df.head(5)
```

```python
## Change Data Type
df["Price_new"] = df["Price"].astype(float)
df["Price_new"] = df["Price"].fillna(df["Price"].mean().astype(float))

df["New_price"] = df["Price"].apply(lambda x:x*2)
df.head()
```

```python
## Data Aggrigating And Groping
grouped_mean = df.groupby("Product")["Price"].mean()
grouped_mean

grouped_sum = df.groupby(["Product","Region"])["Price"].sum()
grouped_sum

grouped_mean = df.groupby(["Product","Region"])["Price"].sum()
grouped_mean

### Multiple Aggrigate Function
grouped_agg = df.groupby("Region")["Price"].agg(["mean","sum","count"])
grouped_agg
```

```python
### Merging and Joining DataFrames
## Create Sample DataFrames
df1 = pd.DataFrame({"key" : ["A","B","C","D"],
                    "Value1" : [1,2,3,4]})
df2 = pd.DataFrame({"key" : ["A","B","C","D"],
                    "Value2" : [5,6,7,8]})

# Merge DataFrame an the "Key Columns"
pd.merge(df1,df2,on = "key", how = "inner")
```
