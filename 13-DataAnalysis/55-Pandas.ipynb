```python
## Create a DataFrame from a dictionary of List
data = {
    "Name" : ["Raunak","Rish","Pratap","Harsh"],
    "Age" : [30,35,39,38],
    "City" : ["Praj","Alld","Knpr","Prtp"]
}

df = pd.DataFrame(data)
print(df)
print(type(df))
```

```python
## Create a DataFrame From List Of Dictionaries

data = [
    {"Name" : "Raunak","Age" : 42,"City" : "Praj"},
    {"Name" : "Rish","Age" : 54,"City" : "Alld"},
    {"Name" : "Pratp","Age" : 44,"City" : "Knpr"},
    {"Name" : "Harsh","Age" : 49,"City" : "Prtp"}
]

df = pd.DataFrame(data)
df

# Accessing Data From DataFrame
print(df.loc[0,"Name"])
print(df.iloc[0,1])
print(df.iloc[0,2])
print(df.at[1,"Age"])
print(df.iat[2,1])

## Data Manipulation With Data Frame
df["Salary"] = [500000,600000,700000,300000]
df

## Deleting(Removing) a Column
df.drop("Salary",axis = 1) ## Temprary
df
df.drop("Salary",axis = 1,inplace = True) ## Permanently
df

## Add age to the Column
df["Age"] = df["Age"] + 1 ## Every Age Increase By 1
df

df.drop(0,inplace = True)
df
```
